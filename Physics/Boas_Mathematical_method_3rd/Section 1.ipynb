# Section 1: The Geometric Series

General equation to compute the sum of a infinite series

<div align="center">

$ S_n = \frac{a(1 - r^n)}{1-r} $

</div>

By taking limit we can calculated the sum of convergent series


<div align="center">

$ S =  \lim_{n \to \infty} S_n $

$ S = \frac{a}{1-r} $

</div>
