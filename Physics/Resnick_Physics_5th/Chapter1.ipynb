## 3. microncentury

```python
import pint

unit = pint.UnitRegistry()

unit.define("century = 100 * year")

lecture = 50 * unit.minute
microcentury = 1e-6 * unit.century

ratio = (lecture / microcentury).to_reduced_units()

print(round(ratio.magnitude * 100, 2), "%")
```

## 5. Seconds in a year

%error = (abs.(approximate - actual) / actual ) * 100%

```python
import math

time = 1 * unit.years
time = time.to(unit.second)

number = math.pi * 10**7 * unit.seconds
difference = abs(number - time)
ratio = difference / time 
percentage = ratio * 100
print(round(percentage.magnitude, 3), "%")
```

## 32. Planck time 33. Planck mass

```python
import numpy as np

G = 6.67e-11 * unit.meter**3 / (unit.kilogram * unit.second**2)
h = 6.63e-34 * unit.kilogram * unit.meter**2 / unit.second
c = 3.00e8 * unit.meter / unit.second

constants = [c, G, h]

# Rows   = Mass, Length, Time
# Column = c, G, h
dimensions = np.array([
    [
        quantity.dimensionality.get("[mass]", 0),
        quantity.dimensionality.get("[length]", 0),
        quantity.dimensionality.get("[time]", 0)
    ]
    for quantity in constants
]).T

print("Dimension matrix:")
print(dimensions)


#    M^0 L^0 T^1
target_time = np.array([0, 0, 1])

a, b, d = np.linalg.solve(dimensions, target_time)

planck_time = c**a * G**b * h**d

print("\nPlanck Time")
print(f"Exponents: c={a}, G={b}, h={d}")
print(f"Result: {planck_time:.3e}")






#    M^1 L^0 T^0
target_mass = np.array([1, 0, 0])

a, b, d = np.linalg.solve(dimensions, target_mass)

planck_mass = c**a * G**b * h**d

print("\nPlanck Mass")
print(f"Exponents: c={a}, G={b}, h={d}")
print(f"Result: {planck_mass:.3e}")
```
